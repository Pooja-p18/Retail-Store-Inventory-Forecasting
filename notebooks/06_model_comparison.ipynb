```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
results = pd.DataFrame({
    "Model" : [
        "Linear Regression",
        "Random Forest"
    ],
    "MAE" : [
        65.13,
        62.76
    ],
    "RMSE" : [
        82.53,  
        81.17
    ],
    "R2 Score" : [
        0.4186,
        0.4377
    ]
})

results
```

```python
plt.figure(figsize=(10,6))

plt.bar(results["Model"], results["MAE"], color=["blue", "orange"])



plt.title("Model Comparison - Mean Absolute Error")
plt.ylabel("Mean Absolute Error")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.bar(results["Model"], results["RMSE"], color=["blue", "orange"])

plt.title("Model Comparison - Root Mean Squared Error")
plt.ylabel("Root Mean Squared Error")
plt.show()
```

```python
plt.figure(figsize=(10,6))
plt.bar(results["Model"], results["R2 Score"], color=["blue", "orange"])
plt.title("Model Comparison - R² Score")
plt.ylabel("R² Score")
plt.show()
```

```python
best_model = results.loc[results["R2 Score"].idxmax()]

best_model
```
